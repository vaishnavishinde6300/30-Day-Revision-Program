## Track A interview questions

A1 : What is the difference between mutable and immutable objects? Why does it matter? Model answer: Immutable objects (int, float, str, tuple, frozenset) cannot change after creation; any 'change' builds a new object. Mutable objects (list, dict, set) change in place, so every name pointing at them sees the change. It matters for aliasing bugs, default arguments, and hashing: only immutable objects can be dict keys or set members. Go one layer deeper: The classic bug is def f(x, acc=[]): the default list is created once and shared across calls. Use acc=None and create it inside.

A2 : is vs ==? Model answer: == calls __eq__ and compares values; is checks whether both names point to the same object. Use is only for singletons like None. Ajit Reddy Go one layer deeper: Small integers (−5 to 256) and some strings are cached by CPython, so is can appear to work for them. That is an implementation detail; code that depends on it is a bug.

A3 : Why is 0.1 + 0.2 != 0.3, and how do you handle money? Model answer: Floats are binary fractions; 0.1 has no exact binary representation, so tiny errors appear. Compare floats with math.isclose. For money use Decimal (built from strings) with an explicit rounding mode, or store integer paise. Go one layer deeper: Also mention round() is banker's rounding, so financial code should use Decimal.quantize(..., ROUND_HALF_UP) or whatever the regulator specifies.

A4 : Why are strings immutable in Python? Model answer: So they can be hashed and used as dict keys safely, shared between names without copying, and interned. The cost is that building a string piece by piece creates new objects. Go one layer deeper: Hence ''.join(parts) is the idiom: it computes the total length once and copies each piece once, O(n), while repeated += can be O(n²).

A5 : str() vs repr()? Model answer: str is the readable form for users; repr is the unambiguous form for developers, ideally valid Python. repr('5') is "'5'", which tells you it is a string, not a number. Go one layer deeper: In logs and f-strings use !r for values you are debugging: it reveals stray spaces and types.

A6 : What does s[::-1] do and what is its complexity? Model answer: A slice with step −1, which returns a reversed copy. Time and extra space are both O(n) because slicing copies.
