### VARIABLES

A variable is a named container that stores a value. 

You create one by assigning with =.

Python is dynamically typed: you never declare the type in advance; Python figures it out from the value.

```python
my_name = "Vaishnavi Shinde"  #string
my_salary = 120000            #int
my_mood = "Study mood"        #string

print("My name :", my_name)
print("Salary :", my_salary)
print("Mood :", my_mood)
```

```python
TAX_RATE = 0.10
basic_salary = 50000
tax_amount = TAX_RATE * basic_salary

net_salary = basic_salary - tax_amount

print("TAX :",tax_amount)
print("NET SALARY :",net_salary)
```

## type Conversion

Two types of type conversion
1) implicit type conversion - it automatically change the type of data

2) explicit type conversion - we fourcefully convert the type of data

```python
#implicit type conversion

a = 10      #int
b = 11.11   #float
c = a + b   #implicit type conversion it automatically convert into int-to-float

print(type(a), type(b), type(c))

#explicit type conversion
print(f"Before {type(a)} - {float(a)} - After {type(float(a))}")
print(f"Before {type(b)} - {int(b)} - After {type(int(b))}")
print(f"Before {type("505")} - {int(a)} - After {type(int(a))}")
print(f"Before {type(True)} - {float(True)} - After {type(float(True))}")
```

## Numbers: the money trap
#### Floats are binart approximations 

```python
print(0.1 + 0.2, 0.1 + 0.2 == 0.3)   # == check exact value

import math
print(math.isclose(0.1 + 0.2, 0.3)) # isclose checks approximate equality

# Decimal: Used for more precise decimal calculations.
# ROUND_HALF_UP: Rounds a number to the nearest value, with halfway cases rounded up.

from decimal import Decimal, ROUND_HALF_UP
emi = Decimal("12345.675") # Decimal create a decimal number
print(emi)
print(emi.quantize(Decimal("0.01"), rounding=ROUND_HALF_UP))

# quantize() controls the number of decimal places.
# Decimal("0.01") means we want two decimal places.
# ROUND_HALF_UP means halfway values round up.


print(round(2.5), round(3.5), round(2.675, 2)) ## banker's rounding + float error

print(7 // 2, -7 // 2, 7 % 3, -7 % 3, divmod(17, 5))
print(2 ** 100)
```


Float     ----------------------   Decimal values may have precision errors

==	      ----------------------   Exact equality

math.isclose()------------------    Approximate equality

Decimal	----------------------      Accurate decimal calculations

round() ----------------------	    Rounds to nearest even on halfway cases

//	----------------------          Floor division

% ----------------------	        Remainder

divmod()  ----------------------	Quotient + remainder

**	----------------------          Power

##  f-strings: formatting without looking it up

```python
name, amount, rate, months = "Asha", 1250000, 0.1075, 60 
print(f"{name} applied for Rs {amount:,} at {rate:.2%} for {months} months") 
print(f"|{name:<8}|{name:>8}|{name:^8}|") 
print(f"{amount:>15,.2f}") 
print(f"{3.14159:08.3f} {255:x} {255:#b} {7:03d}") 
print(f"{amount=}  {rate * 100 = :.2f}") 
print(f"{name!r} vs {name!s}") 
width = 10 
print(f"{'EMI':>{width}}|") 
from datetime import date 
print(f"{date(2026, 11, 20):%d %b %Y}")
```

## Three string patterns every coding round expects
#### Clean → count → compare. All are O(n).

```python
from collections import Counter 
def is_palindrome(s: str) -> bool: 
    clean = [c.lower() for c in s if c.isalnum()] 
    return clean == clean[::-1] 
def is_anagram(a: str, b: str) -> bool: 
    norm = lambda x: Counter(x.replace(" ", "").lower()) 
    return norm(a) == norm(b) 
def first_unique(s: str) -> str | None: 
    counts = Counter(s) 
    return next((ch for ch in s if counts[ch] == 1), None) 
print(is_palindrome("A man, a plan, a canal: Panama")) 
print(is_anagram("Listen", "Silent")) 
print(first_unique("loanassist"), first_unique("aabb"))
```
